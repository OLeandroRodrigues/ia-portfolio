## Environment / paths / ergonomics bootstrap

```python
# --- environment / paths / ergonomics bootstrap ---
%load_ext autoreload
%autoreload 2

from pathlib import Path
import sys
import pandas as pd
import numpy as np

# Detect project root (assuming this notebook is in repo root or notebooks/)
_here = Path().resolve()
if (_here / "src").exists():
    PROJ = _here
elif (_here.parent / "src").exists():
    PROJ = _here.parent
else:
    PROJ = _here  # fallback

SRC  = PROJ / "src"
DATA = PROJ / "data"
DATA_RAW       = DATA / "raw"
DATA_INTERIM   = DATA / "interim"
DATA_PROCESSED = DATA / "processed"
ARTIFACTS      = PROJ / "artifacts"
ART_FEATURES   = ARTIFACTS / "features"
ART_MODELS     = ARTIFACTS / "models"
ART_REPORTS    = ARTIFACTS / "reports"

# Add src/ to PYTHONPATH
if str(SRC) not in sys.path:
    sys.path.append(str(SRC))

# Create essential folders (no error if they already exist)
for p in [DATA_RAW, DATA_INTERIM, DATA_PROCESSED, ART_FEATURES, ART_MODELS, ART_REPORTS]:
    p.mkdir(parents=True, exist_ok=True)

# Display options
pd.set_option("display.max_colwidth", 180)
pd.set_option("display.min_rows", 10)

print(f"PROJ = {PROJ}")
print(f"SRC  = {SRC}")
print(f"DATA = {DATA}")
```

## quick read helpers

```python
# --- quick read helpers ---
from typing import Optional
import pandas as pd

def read_csv_safe(path: Path, **kw) -> Optional[pd.DataFrame]:
    path = Path(path)
    if path.exists():
        return pd.read_csv(path, **kw)
    print(f"File not found: {path}")
    return None

# Example usage (adjust file name if different):
df = read_csv_safe(DATA_PROCESSED / "data-google-reviews-clean.csv")
if df is not None:
    display(df.head())
    print(df.shape)
```

## reproducibility seeds

```python
SEED = 42
try:
    import random, numpy as np
    random.seed(SEED)
    np.random.seed(SEED)
    try:
        import tensorflow as tf
        tf.keras.utils.set_random_seed(SEED)
    except Exception:
        pass
    print(f"Seeds configured with SEED={SEED}")
except Exception as e:
    print("Failed to set seeds:", e)
```
